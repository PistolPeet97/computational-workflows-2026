1. What is nf-core?

It is a global community collaborating to build open-source Nextflow components and pipelines for biomedical data, to get reproducible results with continues supported pipelines.

2. How many pipelines are there currently in nf-core?

157 available pipelines with 101 Released, 43 under development and 13 Archived.

3. Are there any non-bioinformatic pipelines in nf-core?

yes, some for generell informatics pipelines, not specific for bioinformatics.

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A pipeline is a collection of steps to work on data analysis and get reproducible results in the end. It taks the raw data and then runs thru different set up programms and algorithms to get the modified data in the end. They are typicaly used to have a one button push for data analysis and not to have every step ba hand but to automate that process of the data analysis.

5. Why do you think nf-core adheres to strict guidelines?

So that we have a good standard in which the data analysis and the automation is worked on and the results in the end are reproducible at best. Also to make the algorithms easy access and maybe availabe for a long time. It also helps other people to let it run on their own system. So if one piepline works, all of them should work.

6. What are the main features of nf-core pipelines?

If one runs on a system, all of them run on the system. They are all reviewed to adhere to the best practice standard they set, with results in reproducible results.
