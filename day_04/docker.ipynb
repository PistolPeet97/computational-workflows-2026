### What is a container?

Its a self contained-package that bundles a bioinformatics tool, along with its specific dependencies, libraries, and operating system components to be able to let it run on almost any device without long integration of the software.

### Why do we use containers?

To have easy acces to the software, without long installation processes and beeing able to let multiple tools run after each other without having dependencies conflicts, since they are all in their won container. This helps to bulid the pieplines and ensure the reproducability of those.

### Explain the RUN and ENV lines you added to the file

RUN apt-get update && \
    apt-get install -y --no-install-recommends curl cowsay && \
    rm -rf /var/lib/apt/lists/*

1. Run Starts the command
2. apt-get update: gets all the possible packages from the debian source
3. apt-get install: installs all the packages neccesary for curl and cowsay but without the additional stuff to keep it small
4. rm -rf: deletes the package list, because that is not needed

ENV PATH="/usr/games:${PATH}"

Sets the path to /usr/games so it knwos where to look when searching for the programs.
